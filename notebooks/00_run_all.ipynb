Runs notebooks 01 to 04 in order, saves the results back into each notebook, and exports an HTML copy of each to `docs/run_exports/`.

The first run can take a minute or two.

```python
import nbformat
from nbclient import NotebookClient
from nbconvert import HTMLExporter
NB = ROOT / "notebooks"
OUT = ROOT / "docs" / "run_exports"; OUT.mkdir(parents=True, exist_ok=True)
for name in ["01_the_problem_and_cost_model", "02_simulating_production",
             "03_finding_the_cheapest_plan", "04_what_if_assumptions_change"]:
    path = NB / f"{name}.ipynb"
    nb = nbformat.read(path, as_version=4)
    NotebookClient(nb, timeout=600, resources={"metadata": {"path": str(NB)}}).execute()
    nbformat.write(nb, path)
    body, _ = HTMLExporter().from_notebook_node(nb)
    (OUT / f"{name}.html").write_text(body, encoding="utf-8")
    print("done:", name)
```
